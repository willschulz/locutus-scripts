```python
import json
import time
import threading

import websocket
```

```python

# A global list to store all incoming messages
likes = []

def on_message(wsapp, message):
    """
    Callback function that runs whenever a new message arrives.
    Here, we parse the message (JSON) and store it in the likes list.
    """
    try:
        data = json.loads(message)  # parse as JSON if messages are JSON-encoded
    except json.JSONDecodeError:
        data = message  # fallback if not valid JSON
    likes.append(data)

def run_ws():
    """
    Creates a WebSocketApp instance and runs it.
    This function will be run in a background thread so the main thread
    can continue and eventually stop the run after 5 seconds.
    """
    wsapp = websocket.WebSocketApp(
        "wss://jetstream2.us-east.bsky.network/subscribe?wantedCollections=app.bsky.feed.post",
        on_message=on_message
    )
    wsapp.run_forever()

def main():
    # Start the WebSocket in a separate thread
    ws_thread = threading.Thread(target=run_ws, daemon=True)
    ws_thread.start()

    # Give the connection 5 seconds to collect data
    time.sleep(5)

    # After 5 seconds, inspect the data we have collected
    print(f"\nCollected {len(likes)} messages in 5 seconds.")

    # If we have any data, show a sample
    if likes:
        print("\nSample post (first entry):")
        print(likes[0])

    # If you want to stop receiving more data, exit the process here
    # or figure out a clean way to shut down the WebSocket (requires access to the wsapp).
    # For brevity, let's just exit now:
    print("\nDone collecting. Exiting now...")

# if __name__ == "__main__":
#     main()
```
